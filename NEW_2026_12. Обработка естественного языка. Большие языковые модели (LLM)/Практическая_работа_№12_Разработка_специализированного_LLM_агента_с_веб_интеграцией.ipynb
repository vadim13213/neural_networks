```python
import gradio as gr
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline, BitsAndBytesConfig
import bitsandbytes
import torch
import datetime
from tavily import TavilyClient
import os
from huggingface_hub import login
from google.colab import userdata

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import HuggingFaceEmbeddings


os.environ["TAVILY_API_KEY"] = userdata.get('TAVILY_API_KEY')
os.environ["HUGGINGFACE_API_KEY"] = userdata.get('HUGGINGFACE_API_KEY')

login(token=os.environ["HUGGINGFACE_API_KEY"])
```

```python
model_name = "Qwen/Qwen2.5-7B-Instruct"

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True
)

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto",
    torch_dtype=torch.float16,
    quantization_config=bnb_config
)

print("Модель Qwen2.5-7B загружена в 4-bit")
```

```python

tavily = TavilyClient(api_key=os.environ["TAVILY_API_KEY"])

VITEK_PROMPT = """
Ты — Витёк, студент 4 курса, который прошёл через всё: сессии, долги, съёмные квартиры, пустой холодильник и романтику на пару копеек.
Ты общаешься как два студента в курилке — коротко, с юмором, без занудства.

Стиль:
- Используй слова: «кек», «жесть», «забей», «не ссы», «крч», «лады», «ровно», «так себе тема», «всё норм».
- Не пиши длинных абзацев — только по делу.
- Никакой сложной терминологии, только житейские лайфхаки.

Твои навыки:
1. Подсказать, как сдать экзамен, если учил всё ночью.
2. Ответить, что сказать преподу, если не был на парах.
3. Предложить вкусно поесть на 100 рублей (или хотя бы съедобно).
4. Помочь помириться с девушкой/парнем, даже если не прав.
5. Быстро навести порядок, пока не приехали родители.
6. Дать совет по первой работе или стажировке.
7. Объяснить, как пережить похмелье на первой паре.

Всегда отвечай на русском. Не будь занудой.
"""
```

```python

vectorstore = None

def process_pdf(file_path):
    """Загружает PDF, разбивает на чанки, создаёт FAISS индекс."""
    global vectorstore
    try:
        loader = PyPDFLoader(file_path)
        docs = loader.load()
        splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
        chunks = splitter.split_documents(docs)
        embeddings = HuggingFaceEmbeddings(model_name="intfloat/multilingual-e5-small")
        vectorstore = FAISS.from_documents(chunks, embeddings)
        return f"✅ PDF загружен, создано {len(chunks)} фрагментов."
    except Exception as e:
        return f"❌ Ошибка загрузки PDF: {e}"

def retrieve_from_pdf(query, k=3):
    """Ищет релевантные фрагменты в загруженном PDF."""
    global vectorstore
    if vectorstore is None:
        return ""
    docs = vectorstore.similarity_search(query, k=k)
    if not docs:
        return ""
    context = "\n\n---\n\n".join([d.page_content for d in docs])
    return f"📄 Из загруженного PDF:\n{context}"


def search_internet(query: str) -> dict:
    try:
        response = tavily.search(query=query, include_answer=True, include_raw_content=True)
        return {
            "answer": response.get("answer", "Не удалось найти информацию по данному запросу."),
            "sources": [{"url": src["url"], "title": src["title"]} for src in response.get("results", [])[:3]]
        }
    except Exception as e:
        print(f"Ошибка при поиске: {e}")
        return {"answer": "Произошла ошибка при поиске информации.", "sources": []}

def generate_response(message: str, history: list, internet_search: bool = False, use_pdf: bool = False) -> str:

    messages = [{"role": "system", "content": VITEK_PROMPT}]
    for user_msg, bot_msg in history:
        messages.append({"role": "user", "content": user_msg})
        messages.append({"role": "assistant", "content": bot_msg})

    context = ""
    sources = []

    if use_pdf:
        pdf_context = retrieve_from_pdf(message)
        if pdf_context:
            context += f"\n\n{pdf_context}"

    if internet_search:
        search_result = search_internet(message)
        if search_result.get("answer"):
            context += f"\n\n🌐 Интернет:\n{search_result['answer']}"
        sources = search_result.get("sources", [])

    final_message = message + context
    messages.append({"role": "user", "content": final_message})

    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(text, return_tensors="pt").to(model.device)
    outputs = model.generate(
        **inputs,
        max_new_tokens=512,
        temperature=0.7,
        top_p=0.9,
        do_sample=True,
        repetition_penalty=1.1,
        pad_token_id=tokenizer.eos_token_id,
        eos_token_id=tokenizer.eos_token_id
    )
    generated_tokens = outputs[0][inputs.input_ids.shape[1]:]
    response = tokenizer.decode(generated_tokens, skip_special_tokens=True).strip()

    if sources:
        response += "\n\n📚 Источники:\n"
        for src in sources[:3]:
            response += f"- {src['title']}: {src['url']}\n"

    return response
```

```python
def save_chat_to_file(history: list, filename: str = None):
    if filename is None:
        now = datetime.datetime.now().strftime("%Y-%m-%d_%H-%M-%S")
        filename = f"chat_history_{now}.txt"

    with open(filename, "w", encoding="utf-8") as f:
        f.write("История чата с Витьком:\n\n")
        for i, (user_msg, bot_msg) in enumerate(history, 1):
            f.write(f"{i}. Клиент: {user_msg}\n")
            f.write(f"   Витек: {bot_msg}\n\n")

    return filename

```

```python



def chat_interface(message: str, history: list, use_internet: bool, use_pdf: bool) -> str:
    response = generate_response(message, history, internet_search=use_internet, use_pdf=use_pdf)
    return response

with gr.Blocks(title="Витек помощник") as demo:
    gr.Markdown("Витёк – твой бро по выживанию в универе")

    with gr.Row():
        with gr.Column(scale=4):
            chatbot = gr.Chatbot(height=750, label="Чат с Витьком")
            message = gr.Textbox(label="Ваш вопрос", placeholder="Спросите о том, что вас интересует")
            with gr.Row():
                submit = gr.Button("Отправить вопрос")
                clear = gr.Button("Очистить чат")
                internet_toggle = gr.Checkbox(label="🌐 Искать в интернете", value=False)
                save_btn = gr.Button("💾 Сохранить чат")
        with gr.Column(scale=1):
            pdf_upload = gr.File(label="📄 Загрузить PDF", file_types=[".pdf"])
            pdf_status = gr.Textbox(label="Статус PDF", interactive=False)
            pdf_cb = gr.Checkbox(label="Использовать загруженный PDF", value=False)

    save_output = gr.Textbox(label="Сохранение чата", visible=False)
    save_success = gr.Markdown("", visible=False)

    def user(user_message, history):
        return "", history + [[user_message, None]]

    def bot(history, internet, pdf):
        bot_message = chat_interface(history[-1][0], history[:-1], internet, pdf)
        history[-1][1] = bot_message
        return history

    def save_chat(history):
        if not history:
            return {save_success: gr.Markdown("Нет истории для сохранения", visible=True)}
        filename = save_chat_to_file(history)
        return {save_success: gr.Markdown(f"Чат сохранен в файл: {filename}", visible=True)}

    message.submit(user, [message, chatbot], [message, chatbot], queue=False).then(
        bot, [chatbot, internet_toggle, pdf_cb], chatbot
    )
    submit.click(user, [message, chatbot], [message, chatbot], queue=False).then(
        bot, [chatbot, internet_toggle, pdf_cb], chatbot
    )
    clear.click(lambda: None, None, chatbot, queue=False)
    save_btn.click(save_chat, inputs=[chatbot], outputs=[save_success])

    pdf_upload.change(process_pdf, inputs=pdf_upload, outputs=pdf_status)

demo.launch(debug=True)
```
